## Metrik

**Hauptmetrik: F2**, mit „entwickelt innerhalb von fünf Jahren Diabetes“ (`Outcome = 1`) als positiver Klasse, und einer **Precision-Untergrenze `P_min = 0.5`**.

- **Warum ein übersehener Fall schlimmer ist:** `Outcome = 1` heißt, dass Diabetes 1 bis 5 Jahre *nach* der Untersuchung diagnostiziert wurde. Das Modell stellt also eine Prognose, keine Diagnose. Ein falsch Negativer ist eine verpasste Chance zur Vorbeugung (Beratung zum Lebensstil, engmaschige Kontrolle). Ein falsch Positiver kostet eine unnötige Kontrolle und etwas Sorge.
- **Warum β = 2:** F-beta gewichtet einen falsch Negativen β²-mal so stark wie einen falsch Positiven. Wir nehmen an, dass ein übersehener Fall etwa so viel wiegt wie vier Fehlalarme, also β = 2. Das ist eine Annahme, keine Tatsache.
- **Warum eine Precision-Untergrenze:** F2 belohnt es, zu viele Patientinnen als positiv vorherzusagen. `P_min = 0.5` heißt: höchstens eine unnötige Kontrolle pro gefundenem späteren Fall. Nach dem Split prüfen wir, dass `P_min` deutlich über dem Anteil positiver Fälle im Trainings-Set liegt. Dieser Anteil ist die Precision von „alle haben Diabetes“.
- **Wie wir sie einsetzen:** Wir wählen die Entscheidungsschwelle auf den Trainingsdaten so, dass F2 möglichst hoch ist und die Precision mindestens `P_min` bleibt. Accuracy steht in der Ergebnistabelle nur als Warnlampe.

### 4.2 Wie sind die Werte verteilt?

Die Schiefe entscheidet, wie wir fehlende Werte auffüllen (Median bei schiefen Spalten), und zusammen mit den Histogrammen und dem Shapiro-Test, welchen Test wir für H1 und H3 verwenden. Faustregel: |Schiefe| unter 0,5 ist ungefähr symmetrisch, über 1 stark schief.

```python
FEATURES = df_train.columns.drop("Outcome")

df_train[FEATURES].skew().round(2).sort_values(ascending=False)
```

```python
HYPOTHESIS_COLUMNS = ["BMI", "Pregnancies", "Insulin"]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, HYPOTHESIS_COLUMNS):
    sns.histplot(
        data=df_train, x=col, hue="Outcome", stat="density", common_norm=False,
        discrete=(col == "Pregnancies"), kde=(col != "Pregnancies"), ax=ax,
    )
    ax.set_title(f"{col} nach Outcome (Training, ohne NaN)")
plt.tight_layout()
plt.show()
```

```python
# Shapiro je Gruppe: Bei n im Hunderterbereich zusammen mit Histogramm und Schiefe lesen
rows = []
for col in ["BMI", "Insulin"]:
    for outcome, values in df_train.groupby("Outcome")[col]:
        values = values.dropna()
        w, p = stats.shapiro(values)
        rows.append({"Spalte": col, "Outcome": outcome, "n": len(values),
                     "Schiefe": round(values.skew(), 2), "Shapiro W": round(w, 3), "Shapiro p": p})
pd.DataFrame(rows)
```

**Was wir sehen:**

- **Schiefe:** Insulin (2,00), DiabetesPedigreeFunction (1,82) und Age (1,13) sind stark rechtsschief: viele kleine Werte und ein langer Ausläufer großer Werte. Pregnancies, SkinThickness, BMI und Glucose sind mäßig schief (0,55 bis 0,89), BloodPressure ist fast symmetrisch (0,11).
- **Shapiro** verwirft die Normalverteilung für BMI und Insulin in beiden Gruppen (alle p < 0,001). Bei 100 bis 400 Werten je Gruppe reagiert Shapiro schon auf kleine Abweichungen, deshalb lesen wir ihn zusammen mit den Histogrammen: BMI sieht fast glockenförmig aus, mit einem rechten Ausläufer in der Diabetes-Gruppe (Schiefe 0,95). Insulin ist deutlich schief (2,69 und 1,52), mit einem kleinen zweiten Buckel um 500 in der Diabetes-Gruppe.
- **Folge für die Tests:** Unsere Regel aus Schritt 1 lautet „Shapiro verworfen → Mann-Whitney U“, also verwenden wir Mann-Whitney U für H1 und H3. Für BMI wäre wegen der großen Stichprobe auch Welch vertretbar, aber nach dem Datenblick zu wechseln, würde unsere eigene Regel verbiegen. H2 braucht keinen Shapiro: Pregnancies ist eine Zählvariable und geht in Bändern in den χ²-Test.
- **Folge für die Bereinigung:** Weil die meisten Spalten schief sind, füllen wir fehlende Werte mit dem Median auf, nicht mit dem Mittelwert (4.6).

### 4.4 Korrelationen

Nach der Tabelle „two variables at a time“ aus W2D1:

- **Merkmal × Merkmal** (beide metrisch): **Pearsons r** misst, wie gut eine Gerade passt. **Spearmans ρ** ist Pearsons r auf den Rängen. Es misst also jeden monotonen Zusammenhang („je mehr …, desto mehr …“) und ist weniger empfindlich für Schiefe und Ausreißer. Beide reichen von −1 bis +1.
- **Merkmal × `Outcome`** (metrisch × Kategorie): das **Korrelationsverhältnis η**, also wie viel der Streuung einer Spalte die beiden `Outcome`-Gruppen erklären. Bereich 0 bis 1, ohne Richtung.

Wo Pearson und Spearman deutlich voneinander abweichen, ist der Zusammenhang gekrümmt oder wird von wenigen Extremwerten getragen. All das sind Beschreibungen, keine Tests: Sie zählen nicht zu unseren drei Tests. NaN werden paarweise weggelassen.

```python
pearson = df_train[FEATURES].corr(method="pearson")
spearman = df_train[FEATURES].corr(method="spearman")

# Nur die untere Hälfte zeigen (die obere spiegelt sie); leere erste Zeile und letzte Spalte weglassen
mask = np.triu(np.ones((len(FEATURES) - 1, len(FEATURES) - 1), dtype=bool), k=1)
fig, axes = plt.subplots(1, 2, figsize=(17, 7))
for ax, (name, matrix) in zip(axes, [("Pearson's r", pearson), ("Spearman's ρ", spearman)]):
    sns.heatmap(matrix.iloc[1:, :-1], mask=mask, annot=True, fmt=".2f", cmap="vlag",
                vmin=-1, vmax=1, square=True, ax=ax)
    ax.set_title(f"{name} zwischen den Merkmalen (Training)")
plt.tight_layout()
plt.show()
```

```python
# Welche Paare beurteilen die beiden Verfahren am unterschiedlichsten?
pairs = []
for i, a in enumerate(FEATURES):
    for b in FEATURES[i + 1:]:
        pairs.append({"Paar": f"{a} × {b}", "Pearson r": pearson.loc[a, b],
                      "Spearman ρ": spearman.loc[a, b]})
pairs = pd.DataFrame(pairs).set_index("Paar")
pairs["Differenz ρ − r"] = pairs["Spearman ρ"] - pairs["Pearson r"]
pairs.reindex(pairs["Differenz ρ − r"].abs().sort_values(ascending=False).index).head(6).round(2)
```

```python
def correlation_ratio(values, groups):
    """η: Anteil der Streuung in `values`, den die Gruppen erklären, als Wurzel (0 bis 1)."""
    data = pd.DataFrame({"values": values, "groups": groups}).dropna()
    overall_mean = data["values"].mean()
    group_stats = data.groupby("groups")["values"].agg(["count", "mean"])
    ss_between = (group_stats["count"] * (group_stats["mean"] - overall_mean) ** 2).sum()
    ss_total = ((data["values"] - overall_mean) ** 2).sum()
    return np.sqrt(ss_between / ss_total)


# Jede Spalte gegen Outcome, mit allen drei Maßen
with_outcome = pd.DataFrame({
    "Pearson r": df_train[FEATURES].corrwith(df_train["Outcome"], method="pearson"),
    "Spearman ρ": df_train[FEATURES].corrwith(df_train["Outcome"], method="spearman"),
    "η": {col: correlation_ratio(df_train[col], df_train["Outcome"]) for col in FEATURES},
})
with_outcome.sort_values("η", ascending=False).round(3)
```

**Was wir sehen:**

*Merkmal × Merkmal:*
- **Pearson und Spearman stimmen eng überein:** Die größte Abweichung beträgt 0,08 (BMI × Age). Spearman liegt etwas höher, wo eine schiefe Spalte beteiligt ist (Insulin, Age), weil Ränge Extremwerte dämpfen. Gekrümmte Zusammenhänge, die nur Spearman sehen würde, gibt es nicht.
- **Drei starke Paare:**
  - **Pregnancies × Age (ρ = 0,61):** Ältere Frauen hatten mehr Schwangerschaften. Für H2 heißt das: Ein einfacher Test kann nicht unterscheiden, ob die Schwangerschaften oder das Alter den Effekt treiben.
  - **Glucose × Insulin (ρ = 0,67):** Beide stammen aus demselben Glukosetoleranztest. Ein Teil eines Insulin-Effekts kann der Glucose-Effekt sein.
  - **SkinThickness × BMI (ρ = 0,68):** Beide messen Körperfett und tragen also teilweise dieselbe Information. Ein weiterer Grund für die Modellvariante ohne SkinThickness.
- **Kein Paar über 0,7:** keine ernste Multikollinearität für die logistische Regression, aber wir behalten diese Paare im Kopf, wenn wir ihre Koeffizienten lesen.

*Merkmal × `Outcome`:*
- **η ist exakt gleich Pearsons r.** Das ist kein Zufall: Bei zwei Gruppen ist η dasselbe wie der Betrag von Pearsons r mit einer 0/1-Spalte (die punktbiseriale Korrelation). η bringt hier also nichts Neues.
- **Reihenfolge:** Glucose ≈ 0,5, dann Insulin, BMI und SkinThickness (etwa 0,3 bis 0,4), der Rest 0,3 oder darunter.
- **Insulin:** Spearman (0,42) liegt deutlich über Pearson (0,35), weil die Ausreißer Pearson drücken. Das stützt den rangbasierten Mann-Whitney-U-Test für H3.
- **Grenzen:** Jede Zahl betrachtet ein Merkmal für sich. Wie viel ein Merkmal *zusätzlich* zu den anderen beiträgt, zeigt erst die logistische Regression. Und diese Zahlen sind Beschreibungen, keine Tests, sie zählen also nicht gegen α.

### 4.5 Die drei Hypothesen testen

**Was ein Test macht:** Er geht von der Nullhypothese H0 aus, „es gibt keinen Unterschied zwischen den Gruppen“, und fragt, wie wahrscheinlich ein Ergebnis, das mindestens so extrem ist wie unseres, wäre, wenn H0 stimmte. Diese Wahrscheinlichkeit ist der p-Wert. Ein kleiner p-Wert heißt: Unsere Daten passen schlecht zu H0, also verwerfen wir sie. Ein großer p-Wert beweist H0 **nicht** („nicht verworfen ≠ angenommen“), er heißt nur, dass die Daten nicht eindeutig genug sind.

**Signifikanzniveau:** Wir führen drei Tests durch. Mit α = 0,05 je Test läge die Wahrscheinlichkeit für mindestens einen Fehlalarm bei etwa 1 − 0,95³ ≈ 14 %. Bonferroni teilt α durch die Zahl der Tests: **α = 0,05 / 3 ≈ 0,0167**. Das wurde in Schritt 1 festgelegt.

**p-Wert und Effektstärke gehören zusammen:** Der p-Wert sagt, ob ein Unterschied wahrscheinlich echt ist, die Effektstärke, wie groß er ist. Bei vielen Patientinnen wird selbst ein winziger Unterschied signifikant, deshalb berichten wir beides.

**Welcher Test und warum** (aus dem Testplan in Schritt 1 und den Verteilungen in 4.2):

| | Spalte | Test | Warum |
| :-- | :-- | :-- | :-- |
| H1 | `BMI` | Mann-Whitney U, einseitig (Outcome 1 größer) | Shapiro verwirft die Normalverteilung in beiden Gruppen (4.2), also nehmen wir den rangbasierten Test. Die Hypothese hat eine Richtung. |
| H2 | `Pregnancies` in Bändern 0 / 1–3 / 4–6 / 7+ | χ²-Unabhängigkeitstest | Zwei kategoriale Variablen (Band × `Outcome`). |
| H3 | `Insulin` | Mann-Whitney U, zweiseitig | Stark schief (Schiefe 2,69 und 1,52), und die Richtung ist offen. |

**Mann-Whitney U in einem Satz:** Er bringt alle Werte beider Gruppen in eine gemeinsame Rangfolge und prüft, ob eine Gruppe in dieser Rangfolge tendenziell höher liegt als die andere. Er vergleicht Ränge, keine Mittelwerte, deshalb können wenige Extremwerte ihn nicht dominieren. Seine Effektstärke ist die **rangbiseriale Korrelation** r = 2 · U / (n₁ · n₂) − 1: +1 heißt, jede Patientin mit Diabetes hat einen höheren Wert als jede ohne, 0 heißt keine Tendenz, −1 das Gegenteil.

```python
ALPHA = 0.05 / 3


def mann_whitney(col, alternative):
    """Mann-Whitney U zwischen Outcome 1 und Outcome 0 auf dem Trainings-Set, NaN weggelassen."""
    with_diabetes = df_train.loc[df_train["Outcome"] == 1, col].dropna()
    without_diabetes = df_train.loc[df_train["Outcome"] == 0, col].dropna()
    u, p = stats.mannwhitneyu(with_diabetes, without_diabetes, alternative=alternative)
    rank_biserial = 2 * u / (len(with_diabetes) * len(without_diabetes)) - 1
    return {
        "Spalte": col, "Test": f"Mann-Whitney U ({alternative})",
        "n Outcome 1": len(with_diabetes), "n Outcome 0": len(without_diabetes),
        "Median Outcome 1": with_diabetes.median(), "Median Outcome 0": without_diabetes.median(),
        "U": u, "p": p, "Effektstärke": f"rangbiseriale r = {rank_biserial:.2f}",
        "signifikant bei α = 0,0167": p < ALPHA,
    }
```

#### H1: BMI

H0: Die BMI-Werte der Frauen, die Diabetes entwickeln, liegen nicht tendenziell höher als die der Frauen, die keinen entwickeln. H1 (einseitig): Sie liegen tendenziell höher. Plot: das BMI-Feld in 4.2 und der BMI-Boxplot in 4.3.

```python
h1 = mann_whitney("BMI", alternative="greater")
pd.Series({**h1, "p": f"{h1['p']:.2e}"})
```

**Was wir sehen:** p = 1,6e-15 liegt weit unter α = 0,0167, also verwerfen wir H0. Der Median-BMI liegt in der Diabetes-Gruppe bei 34,35, in der anderen Gruppe bei 30,1.

**So liest man die Effektstärke:** rangbiseriale Korrelation r = 0,39. Als Wahrscheinlichkeit: (r + 1) / 2 = 0,70. Ziehen wir eine Frau, die Diabetes entwickelt, und eine, die keinen entwickelt, hat die erste in **70 % der Paare** den höheren BMI (50 % hieße kein Zusammenhang). Ein mittlerer Effekt.

**Was genau getestet wurde:** dass die BMI-Werte der Diabetes-Gruppe *tendenziell höher liegen* (ihre Ränge), nicht, dass der Mittelwert höher ist. → **H1 gestützt.**

#### H2: Pregnancies

Der χ²-Test vergleicht die beobachteten Zahlen in jeder Zelle der Tabelle Band × `Outcome` mit den Zahlen, die wir erwarten würden, wenn der Diabetes-Anteil in jedem Band gleich wäre. Je größer die Abweichung, desto größer χ² und desto kleiner p. Voraussetzung: mindestens 5 erwartete Fälle pro Zelle, das prüfen wir.

**Einschränkung:** χ² fragt nur, *ob* sich der Diabetes-Anteil zwischen den Bändern unterscheidet, nicht *in welche Richtung*. Unsere Hypothese sagt „je mehr Schwangerschaften, desto wahrscheinlicher“. Die Richtung lesen wir deshalb am Diabetes-Anteil je Band ab, und der Verdict zählt nur dann als „gestützt“, wenn die Anteile mit den Bändern steigen.

**Effektstärke: Cramér's V**, von 0 (kein Zusammenhang) bis 1 (das Band bestimmt das Outcome vollständig). Grobe Orientierung für eine Tabelle mit zwei Spalten: 0,1 klein, 0,3 mittel, 0,5 groß.

```python
PREGNANCY_BANDS = [-0.5, 0.5, 3.5, 6.5, np.inf]
PREGNANCY_LABELS = ["0", "1–3", "4–6", "7+"]

pregnancy_band = pd.cut(df_train["Pregnancies"], bins=PREGNANCY_BANDS, labels=PREGNANCY_LABELS)
observed = pd.crosstab(pregnancy_band, df_train["Outcome"])

chi2, p, dof, expected = stats.chi2_contingency(observed)
cramers_v = np.sqrt(chi2 / (observed.to_numpy().sum() * (min(observed.shape) - 1)))

band_table = observed.copy()
band_table.columns = ["Outcome 0", "Outcome 1"]
band_table["Anteil mit Diabetes"] = (observed[1] / observed.sum(axis=1)).round(3)
band_table["kleinste erwartete Zahl"] = expected.min(axis=1).round(1)
display(band_table)

h2 = {
    "Spalte": "Pregnancies (Bänder)", "Test": "χ²-Unabhängigkeitstest",
    "chi²": chi2, "Freiheitsgrade": dof, "p": p,
    "Effektstärke": f"Cramér's V = {cramers_v:.2f}", "signifikant bei α = 0,0167": p < ALPHA,
}
pd.Series({**h2, "p": f"{h2['p']:.2e}"})
```

```python
ax = band_table["Anteil mit Diabetes"].plot(kind="bar", color="tab:orange", figsize=(6, 4), rot=0)
ax.axhline(p_train, color="grey", linestyle="--", label=f"Anteil im gesamten Trainings-Set ({p_train:.2f})")
ax.set_xlabel("Anzahl der Schwangerschaften (Band)")
ax.set_ylabel("Anteil, der Diabetes entwickelt")
ax.set_title("H2: Anteil mit Diabetes je Schwangerschaftsband (Training)")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Ist das Band 7+ einfach die älteste Gruppe? (Spearman Pregnancies × Age = 0,61, siehe 4.4)
df_train.groupby(pregnancy_band, observed=True)["Age"].agg(["median", "mean"]).round(1)
```

**Was wir sehen:**

- **Voraussetzung erfüllt:** Die kleinste erwartete Zellzahl ist 29,6, weit über 5.
- **χ² = 39,4 bei 3 Freiheitsgraden** ((4 Bänder − 1) × (2 Outcomes − 1)), **p = 1,4e-8**: Der Diabetes-Anteil hängt vom Band ab. **Cramér's V = 0,25**, ein kleiner bis mittlerer Effekt.
- **Aber das Muster ist nicht „je mehr, desto häufiger“:** Band 0 hat 0,37, Band 1–3 nur 0,25, Band 4–6 0,32, und Band 7+ springt auf 0,57. Der Anteil steigt erst ab 4–6.
- **Alter:** Band 7+ ist zugleich die älteste Gruppe (Median-Alter 42, gegenüber 24, 25 und 34). Bei Spearman ρ = 0,61 zwischen Pregnancies und Age kann dieser Test die beiden nicht trennen.
- **Verdict:** Vor dem Test haben wir festgelegt, dass χ² nur sagt, *ob* es einen Zusammenhang gibt, und dass H2 nur dann als gestützt zählt, wenn die Anteile mit den Bändern steigen. Sie steigen nicht durchgehend → **H2 noch nicht entscheidbar.**
- **Band 0 über 1–3** ist eine Überraschung und kommt in „Nebenbei gefunden“. Mit 85 Frauen ist es eine kleine Gruppe: 31 von 85 können leicht um etwa ±10 Prozentpunkte schwanken.

#### H3: Insulin

H0: Die Insulinwerte der beiden Gruppen unterscheiden sich in ihrer Tendenz nicht. H1 (zweiseitig): Sie unterscheiden sich, in die eine oder andere Richtung. Es zählen nur die 324 Trainingszeilen mit gemessenem Insulinwert (4.1: Insulin fehlt in beiden Gruppen etwa gleich oft, 46 % und 49 %). Plot: das Insulin-Feld in 4.2 und der Insulin-Boxplot in 4.3.

```python
h3 = mann_whitney("Insulin", alternative="two-sided")
pd.Series({**h3, "p": f"{h3['p']:.2e}"})
```

**Was wir sehen:** Nur die 324 Zeilen mit gemessenem Insulinwert sind beteiligt (109 mit und 215 ohne Diabetes). p = 6,5e-14, also verwerfen wir H0. Das Median-Insulin liegt in der Diabetes-Gruppe bei 168, in der anderen Gruppe bei 100.

- **Effektstärke:** r = 0,51, also (r + 1) / 2 = 0,76: In drei von vier Paaren hat die Frau, die Diabetes entwickelt, den höheren Insulinwert. Der größte Effekt der drei Tests.
- **Die offene Richtung ist beantwortet: höher.** Das passt zur Insulinresistenz: Der Körper schüttet mehr Insulin aus, um den Blutzucker niedrig zu halten. Das ist das frühe Stadium, also genau das, was wir erwarten würden, wenn Diabetes erst 1 bis 5 Jahre *später* diagnostiziert wird.
- **Einschränkungen:** Der Test nutzt nur 324 von 614 Zeilen (in beiden Gruppen fehlt der Wert etwa gleich oft, 4.1), und Insulin korreliert mit Glucose (ρ = 0,67). Ein Teil des Effekts ist also vermutlich der Glucose-Effekt. → **H3 gestützt, Richtung: höher.**

### 4.6 Bereinigung: fehlende Werte mit den Trainings-Medianen auffüllen

Die Tests oben liefen absichtlich auf der NaN-Version: Hätten wir die Hälfte der Insulin-Spalte mit einem einzigen Wert aufgefüllt, wäre der Unterschied zwischen den Gruppen verwischt. Für das Modell brauchen wir jetzt vollständige Zeilen, denn `LogisticRegression` kann nicht mit NaN umgehen.

**Auffüllen oder droppen?** Die Folien aus W2D1 nennen fünf Möglichkeiten, von den wenigsten bis zu den meisten Annahmen: Zeilen löschen, mit einer Konstanten auffüllen, mit Mittelwert oder Median auffüllen, interpolieren, aus den anderen Spalten vorhersagen. Und sie warnen: „Cleaning decisions are analysis decisions.“ Zeilen löschen verändert, wer in der Stichprobe ist, und Auffüllen verengt die Streuung und zieht Korrelationen Richtung null. Das hieße jede Möglichkeit für unsere Daten:

| Möglichkeit | Was das hier hieße | Entscheidung |
| :-- | :-- | :-- |
| Jede Zeile mit einem NaN löschen | Nur 322 von 614 Trainingszeilen sind vollständig. Fast die Hälfte der Patientinnen wäre weg, und das Modell würde nur von Frauen lernen, die die vollständige Untersuchung hatten. | nein, zu teuer |
| Die Spalten `Insulin` und `SkinThickness` löschen | Alle Zeilen bleiben, aber wir verlieren Insulin, das zweitstärkste Signal (H3: r = 0,51). | als **Variante** in Schritt 6 |
| Mit dem Trainings-Median auffüllen | Zeilen und Spalten bleiben. Für Glucose, BloodPressure und BMI (unter 5 % fehlen) ist das unbedenklich. Für Insulin (47 % fehlen) entsteht ein großer Block bei 125, der den Zusammenhang abschwächt. | **Hauptversion** |
| Eine Spalte „wurde gemessen“ (0/1) ergänzen | Nützlich, wenn schon die Lücke Information trägt (MNAR). Hier hängt das Fehlen kaum von `Outcome` ab (4.1), es gibt also wenig zu gewinnen. | nicht verwendet |

Wir füllen also mit dem Median auf **und** trainieren eine Variante ohne die beiden Spalten. Unterscheiden sich die Scores der beiden Versionen deutlich, spielen die aufgefüllten Werte eine Rolle, und das sagen wir dann („Try the alternative, and see whether the conclusion moves“).

Bei `BloodPressure` (Schiefe 0,11) liegen Median und Mittelwert fast gleich, eine Regel für alle fünf Spalten hält es einfach.

## Schritt 5: Merkmale, Zielvariable und zwei Baselines

**Warum Baselines:** Ein Score allein sagt nichts. F2 = 0,7 klingt gut, aber ist das besser als Raten? Eine Baseline ist das Einfachste, was funktionieren könnte. Ein Modell zählt nur so weit, wie es sie schlägt.

**Die zwei Baselines aus der Aufgabe, plus eine Referenz:**

| | Was sie macht | Warum |
| :-- | :-- | :-- |
| Zufalls-Baseline | `DummyClassifier(strategy="stratified")` rät in 34,9 % der Fälle zufällig „Diabetes“, dem Anteil im Trainings-Set | Die Untergrenze: was man ohne jede Information bekommt |
| Regel-Baseline | „Glucose ≥ Cutoff heißt Diabetes“ | Glucose trennt die Gruppen am deutlichsten (4.3, 4.4). Eine Spalte und eine Zahl: was eine Ärztin mit einem einzigen Laborwert tun könnte |
| Referenz: alle haben Diabetes | Sagt für jede Patientin 1 vorher | Von der Aufgabe nicht verlangt, aber sie zeigt, warum wir `P_MIN` brauchen (siehe unten) |

**Wie wir den Glucose-Cutoff wählen:** mit derselben Regel wie später für das Modell, also der Cutoff mit dem **höchsten F2 unter denen mit Precision ≥ `P_MIN`**, gewählt auf den Trainingsdaten. Nur dann ist der Vergleich zwischen Baseline und Modell fair. Würden wir den Cutoff nach Accuracy wählen und das Modell nach F2 beurteilen, wäre die Baseline von vornherein schwach.

**Welche Daten:** `df_train_clean` aus 4.6 (fehlende Werte mit Trainings-Medianen aufgefüllt). Alle Scores in diesem Schritt sind **Trainingswerte**. Die Testwerte folgen in Schritt 8.

```python
TARGET = "Outcome"
FEATURES_WITHOUT_INSULIN_SKIN = [col for col in FEATURES if col not in ["Insulin", "SkinThickness"]]

X_train = df_train_clean[FEATURES]
y_train = df_train_clean[TARGET]
print("X_train:", X_train.shape, "| y_train:", y_train.shape)
print("Merkmale von Variante B (ohne Insulin und SkinThickness):", FEATURES_WITHOUT_INSULIN_SKIN)
```

**Eine Funktion für alle Scores:** Jedes Modell wird gleich bewertet, damit die Tabelle am Ende Gleiches mit Gleichem vergleicht. F2 ist unsere Hauptmetrik, Precision und Recall zeigen, woraus F2 besteht, Accuracy ist nur eine Warnlampe. `erfüllt P_min` sagt, ob das Modell unsere Precision-Untergrenze einhält.

```python
def classification_scores(y_true, y_pred):
    """Die Scores, die wir für jedes Modell berichten: F2 (Hauptmetrik), Precision, Recall, Accuracy."""
    precision = precision_score(y_true, y_pred, zero_division=0)
    return {
        "F2": fbeta_score(y_true, y_pred, beta=2, zero_division=0),
        "precision": precision,
        "recall": recall_score(y_true, y_pred, zero_division=0),
        "accuracy": accuracy_score(y_true, y_pred),
        "erfüllt P_min": precision >= P_MIN,
    }


train_results = {}  # Modellname -> Trainingswerte, Schritt für Schritt gefüllt
```

### 5.1 Zufalls-Baseline

```python
dummy = DummyClassifier(strategy="stratified", random_state=42)
dummy.fit(X_train, y_train)
train_results["Zufalls-Baseline"] = classification_scores(y_train, dummy.predict(X_train))

train_results["Referenz: alle haben Diabetes"] = classification_scores(y_train, np.ones(len(y_train), dtype=int))

pd.DataFrame.from_dict(train_results, orient="index").round(3)
```

**Was wir sehen:**

- **Zufalls-Baseline:** Precision und Recall liegen beide bei etwa 0,34, F2 = 0,34. Das bringt Raten im richtigen Verhältnis: Von den 34,9 % markierten Fällen sind zufällig 34,9 % echte Fälle. Die Accuracy ist 0,54, und das zeigt, warum Accuracy nur eine Warnlampe ist: Sie klingt nach „halb richtig“, aber das Modell weiß nichts. Die Precision-Untergrenze erfüllt es auch nicht.
- **Alle haben Diabetes: F2 = 0,73.** Es findet jeden Fall (Recall = 1,0), und weil F2 den Recall viermal so stark gewichtet wie die Precision, bekommt dieses nutzlose Modell ein hohes F2. Das passt zur Formel aus der Planung: F2 = 5p / (4p + 1) = 0,728 bei p = 0,349.
- **Genau deshalb haben wir `P_MIN` festgelegt:** Seine Precision ist nur 0,349, unter 0,5, also ist `erfüllt P_min` gleich `False`, und es zählt nicht. Ohne die Untergrenze würde F2 allein Modelle belohnen, die fast alle für krank erklären.

### 5.2 Regel-Baseline: Glucose-Cutoff

Wir probieren jeden Glucose-Wert im Trainings-Set als Cutoff aus, berechnen die Scores für „Glucose ≥ Cutoff heißt Diabetes“, behalten nur die Cutoffs mit Precision ≥ `P_MIN` und nehmen den mit dem höchsten F2.

```python
def rule_predict(data, cutoff):
    """Regel-Baseline: 1 (Diabetes), wenn Glucose mindestens beim Cutoff liegt."""
    return (data["Glucose"] >= cutoff).astype(int)


cutoff_table = pd.DataFrame([
    {"cutoff": cutoff, **classification_scores(y_train, rule_predict(X_train, cutoff))}
    for cutoff in np.sort(X_train["Glucose"].unique())
])

allowed = cutoff_table[cutoff_table["erfüllt P_min"]]
GLUCOSE_CUTOFF = allowed.loc[allowed["F2"].idxmax(), "cutoff"]
best_without_floor = cutoff_table.loc[cutoff_table["F2"].idxmax()]

print(f"Gewählter Cutoff (höchstes F2 bei Precision >= {P_MIN}): Glucose >= {GLUCOSE_CUTOFF:.0f}")
print(f"Ohne Untergrenze würde F2 Glucose >= {best_without_floor['cutoff']:.0f} "
      f"wählen (Precision {best_without_floor['precision']:.2f})")

train_results[f"Regel: Glucose >= {GLUCOSE_CUTOFF:.0f}"] = classification_scores(y_train, rule_predict(X_train, GLUCOSE_CUTOFF))
```

```python
fig, ax = plt.subplots(figsize=(9, 4.5))
for score in ["F2", "precision", "recall"]:
    ax.plot(cutoff_table["cutoff"], cutoff_table[score], label=score)
ax.axhline(P_MIN, color="grey", linestyle="--", label=f"P_min = {P_MIN}")
ax.axvline(GLUCOSE_CUTOFF, color="black", linestyle=":", label=f"gewählter Cutoff = {GLUCOSE_CUTOFF:.0f}")
ax.set_xlabel("Glucose-Cutoff (mg/dl)")
ax.set_ylabel("Score auf dem Trainings-Set")
ax.set_title("Regel-Baseline: Scores für jeden Glucose-Cutoff")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
pd.DataFrame.from_dict(train_results, orient="index").round(3)
```

**Was wir sehen:**

- **Die Kurven:** Ein niedriger Cutoff markiert fast alle, also ist der Recall hoch und die Precision niedrig. Ein hoher Cutoff markiert nur eindeutige Fälle, also steigt die Precision und der Recall fällt. F2 erreicht sein Maximum bei einem niedrigen Cutoff, weil es den Recall viermal so stark gewichtet.
- **Gewählter Cutoff: Glucose ≥ 112.** Unter 112 fällt die Precision unter 0,5. Bei 112 findet die Regel 83 % der späteren Fälle (Recall 0,83), und jeder zweite Alarm ist ein echter Fall (Precision 0,50). F2 = 0,73.
- **Ohne die Untergrenze** würde F2 Glucose ≥ 100 wählen: Recall 0,94, aber Precision nur 0,44, also wäre mehr als die Hälfte der Alarme falsch. Die Untergrenze verhindert das.
- **Die Regel schlägt die Zufalls-Baseline deutlich** (F2 0,73 gegen 0,34). Ihr F2 liegt etwa gleichauf mit „alle haben Diabetes“, aber die Regel hält die Precision-Untergrenze ein, die Referenz nicht. Das ist die Messlatte, die die logistische Regression überspringen muss: **F2 über 0,73 bei Precision ≥ 0,5.**
- **Vorsicht:** Der Cutoff wurde auf den Trainingsdaten gewählt und auf denselben Daten bewertet, 0,73 ist also etwas zu optimistisch. Die ehrliche Zahl kommt vom Test-Set in Schritt 8.

## Schritt 6: Logistische Regression

**Was das Modell macht:** Die logistische Regression addiert die Merkmale, jedes mit einem Gewicht (Koeffizient) multipliziert, plus eine Konstante (Intercept). Die Summe sind die Log-Odds für Diabetes. Die Sigmoid-Funktion macht daraus eine Wahrscheinlichkeit zwischen 0 und 1. `predict` sagt dann „Diabetes“, sobald diese Wahrscheinlichkeit mindestens **0,5** beträgt. Diese Schwelle ist ein Standardwert und keine Entscheidung von uns, Schritt 7 wird sie ändern.

**Warum wir zuerst skalieren:** Die Merkmale haben sehr unterschiedliche Einheiten (Glucose um 120, DiabetesPedigreeFunction um 0,5). `StandardScaler` zieht von jeder Spalte den Mittelwert ab und teilt durch ihre Standardabweichung, sodass jedes Merkmal Mittelwert 0 und Standardabweichung 1 hat. Zwei Vorteile: Der Solver konvergiert zuverlässig, und die Koeffizienten werden vergleichbar. Ein Koeffizient bedeutet dann „Änderung der Log-Odds, wenn dieses Merkmal um eine Standardabweichung steigt“.

**Warum wir den Scaler nur auf den Trainingsdaten fitten:** Der Scaler *lernt* einen Mittelwert und eine Standardabweichung. Lernte er sie aus allen Daten, flössen Informationen über die Test-Patientinnen ins Training: Data Leakage, genau wie bei einem aus allen Zeilen berechneten Median. Deshalb fitten wir ihn auf `X_train` und *wenden* ihn auf das Test-Set später nur *an* (`transform`, nicht `fit`).

**Zwei Varianten (aus 4.6):**

| | Merkmale | Warum |
| :-- | :-- | :-- |
| **A** (`logreg`) | alle 8 | die Aufgabe sagt „möglichst viele Merkmale“ |
| **B** (`logreg_b`) | ohne `Insulin` und `SkinThickness` | nach dem Auffüllen sind 47 % und 29 % ihrer Werte derselbe Median: Braucht das Modell sie überhaupt? |

```python
FEATURES_A = list(FEATURES)
FEATURES_B = FEATURES_WITHOUT_INSULIN_SKIN


def fit_logreg(features, class_weight=None):
    """Einen StandardScaler auf den Trainingsmerkmalen fitten, dann eine logistische Regression auf den skalierten Daten."""
    scaler = StandardScaler().fit(df_train_clean[features])
    model = LogisticRegression(class_weight=class_weight)
    model.fit(scaler.transform(df_train_clean[features]), y_train)
    return scaler, model


def predict_probability(scaler, model, data, features):
    """Wahrscheinlichkeit für Diabetes (Klasse 1) für jede Zeile von `data`."""
    return model.predict_proba(scaler.transform(data[features]))[:, 1]


scaler, logreg = fit_logreg(FEATURES_A)
scaler_b, logreg_b = fit_logreg(FEATURES_B)

# predict() verwendet die Standardschwelle 0,5
train_results["logreg A (alle 8 Merkmale)"] = classification_scores(
    y_train, logreg.predict(scaler.transform(df_train_clean[FEATURES_A])))
train_results["logreg B (ohne Insulin, SkinThickness)"] = classification_scores(
    y_train, logreg_b.predict(scaler_b.transform(df_train_clean[FEATURES_B])))

pd.DataFrame.from_dict(train_results, orient="index").round(3)
```

**Was wir sehen:**

- **Variante A: F2 = 0,62, unter der Glucose-Regel (0,73).** Gleichzeitig ist ihre Accuracy (0,80) viel höher als die der Regel (0,65), und ihre Precision (0,77) liegt weit über der Untergrenze. Das Modell ist gut, aber auf das falsche Ziel eingestellt: Mit der Schwelle 0,5 sagt es nur dann „Diabetes“, wenn es sich recht sicher ist, und findet deshalb nur 59 % der späteren Fälle. Unsere Metrik gewichtet einen übersehenen Fall viermal so stark wie einen Fehlalarm, das kostet also viel F2.
- **Das ist die Kernlektion aus W2D4:** Ein Modell ist nicht an sich gut oder schlecht, sondern nur für eine Metrik. Nach Accuracy schlägt das Modell die Regel klar. Nach F2, der Metrik, die wir vor den Ergebnissen gewählt haben, verliert es.
- **Variante B ist praktisch identisch** (F2 0,62, Precision 0,76, Recall 0,59). Insulin und SkinThickness wegzulassen, kostet fast nichts. Warum, zeigen die Koeffizienten unten.
- **Noch nicht geschlagen:** Bisher überspringt kein Modell die Messlatte der Glucose-Regel. Schritt 7 verschiebt Schwelle und Klassengewichte in Richtung unseres Ziels.

### 6.1 Was hat das Modell gelernt? Die Koeffizienten

Weil die Merkmale skaliert sind, lassen sich die Koeffizienten direkt vergleichen. **Odds Ratio** = e^Koeffizient: Mit diesem Faktor werden die Odds für Diabetes multipliziert, wenn das Merkmal um eine Standardabweichung steigt, **während alle anderen Merkmale gleich bleiben**. Ein Odds Ratio von 1 heißt kein Effekt.

```python
coefficients = pd.DataFrame({
    "Koeffizient A": pd.Series(logreg.coef_[0], index=FEATURES_A),
    "Koeffizient B": pd.Series(logreg_b.coef_[0], index=FEATURES_B),
})
coefficients["Odds Ratio A"] = np.exp(coefficients["Koeffizient A"])
coefficients.sort_values("Koeffizient A", ascending=False).round(2)
```

**Was wir sehen:**

- **Glucose dominiert:** Koeffizient 1,18, Odds Ratio 3,3. Eine Standardabweichung mehr Glucose (etwa 30 mg/dl) verdreifacht ungefähr die Odds, Diabetes zu entwickeln, wenn alles andere gleich bleibt. Danach kommen BMI (Odds Ratio 2,0) und Pregnancies (1,5).
- **Insulin: Koeffizient −0,07, SkinThickness: 0,03, praktisch null.** Dabei hatte Insulin in H3 den größten Effekt aller drei Tests (r = 0,51). Das ist kein Widerspruch:
  - Der **Test** hat Insulin **für sich allein** betrachtet. Das **Modell** fragt, was Insulin **zusätzlich** zu den anderen Merkmalen beiträgt. Insulin korreliert mit Glucose (ρ = 0,67, 4.4), der größte Teil seiner Information steckt also schon in Glucose. SkinThickness korreliert genauso mit BMI (ρ = 0,68).
  - Außerdem ist bei der Hälfte der Insulinwerte nur der aufgefüllte Median 125 eingetragen. Diese Zeilen tragen gar keine Information über Insulin.
  - Das erklärt, warum Variante B nichts verliert: Das Modell hat beide Spalten ohnehin ignoriert.
- **Pregnancies (1,5) deutlich über Age (1,2):** Bei gleichem Alter tragen die Schwangerschaften noch etwas bei. Das ist ein Hinweis zu H2, wo der Test Schwangerschaften und Alter nicht trennen konnte. **Es ist ein Hinweis, kein Verdict:** Koeffizienten sind keine Tests, und unsere drei Tests stehen fest. → einen Satz in der Diskussion wert.
- **Vorsicht bei allen Koeffizienten:** Bei korrelierten Merkmalen können sie sich gegenseitig verschieben, und sie beschreiben Zusammenhänge, keine Ursachen.

## Schritt 7: Mit dem Klassenungleichgewicht umgehen

**Das Problem:** 400 der 614 Trainings-Patientinnen (65 %) entwickeln keinen Diabetes. Ein Modell, das seine Fehler minimiert, lernt vor allem von der Mehrheit und wird vorsichtig damit, „Diabetes“ zu sagen. In Schritt 6 hat es nur 59 % der späteren Fälle gefunden.

**Zwei Hebel, die an verschiedenen Stellen ansetzen:**

| | Was er ändert | Wann |
| :-- | :-- | :-- |
| **7a `class_weight="balanced"`** | **was das Modell lernt.** Jede Klasse bekommt das Gewicht n / (2 · n_Klasse): Diabetes-Fälle 614 / (2 · 214) ≈ 1,43, die anderen 614 / (2 · 400) ≈ 0,77. Ein übersehener Diabetes-Fall kostet das Modell beim Training also etwa 1,9-mal so viel, und seine Wahrscheinlichkeiten verschieben sich nach oben. | während `fit` |
| **7b Entscheidungsschwelle** | **wie wir Wahrscheinlichkeiten in Entscheidungen übersetzen.** Statt „Diabetes, wenn p ≥ 0,5“ verwenden wir „Diabetes, wenn p ≥ t“. Das Modell selbst bleibt gleich. | nach `fit`, mit `predict_proba` |

**Wie wir die Schwelle wählen:** genau wie den Glucose-Cutoff in Schritt 5. Wir probieren jede Wahrscheinlichkeit, die im Trainings-Set vorkommt, als Schwelle aus und nehmen die mit dem **höchsten F2 unter denen mit Precision ≥ `P_MIN`**. Wir wählen sie auf den **Trainingsdaten**, nie auf dem Test-Set. Einschränkung: Das Modell hat diese Patientinnen gesehen, die gewählte Schwelle ist also etwas zu optimistisch. In Woche 3 behebt `cross_val_predict` das.

**Hinweis des Coaches (08.10.):** Nach dem Ändern der Klassengewichte die Schwelle neu wählen. Die Gewichte verschieben die Wahrscheinlichkeiten, die alte Schwelle passt also nicht mehr.

### 7a. Klassengewichte: `class_weight="balanced"`

```python
scaler_balanced, logreg_balanced = fit_logreg(FEATURES_A, class_weight="balanced")
scaler_balanced_b, logreg_balanced_b = fit_logreg(FEATURES_B, class_weight="balanced")

train_results["logreg A balanced"] = classification_scores(
    y_train, logreg_balanced.predict(scaler_balanced.transform(df_train_clean[FEATURES_A])))
train_results["logreg B balanced"] = classification_scores(
    y_train, logreg_balanced_b.predict(scaler_balanced_b.transform(df_train_clean[FEATURES_B])))

pd.DataFrame.from_dict(train_results, orient="index").round(3)
```

**Was wir sehen:** Mit den Klassengewichten steigt der Recall von 0,59 auf 0,71 und F2 von 0,62 auf etwa 0,70. Die Precision sinkt von 0,77 auf etwa 0,64, das ist der Preis: mehr Alarme, und mehr davon falsch. Das ist der Zielkonflikt aus W2D4 in Aktion.

Aber **F2 ≈ 0,70 liegt immer noch unter der Glucose-Regel (0,73).** Die Gewichte allein reichen nicht: Das Modell entscheidet weiter bei p ≥ 0,5, und diese Schwelle wurde nicht für unsere Metrik gewählt.

### 7b. Die Entscheidungsschwelle wählen

Wir wählen eine Schwelle für alle vier Modelle (A und B, jeweils ohne und mit Klassengewichten). Die beiden Modelle ohne Gewichte verlangt die Aufgabe nicht, aber sie beantworten eine Frage: **Brauchen wir die Klassengewichte noch, wenn wir die Schwelle selbst wählen?**

```python
def best_threshold(y_true, probability):
    """Höchstes F2 unter den Schwellen mit Precision >= P_MIN; liefert auch die Scores jeder Schwelle."""
    table = pd.DataFrame([
        {"threshold": t, **classification_scores(y_true, (probability >= t).astype(int))}
        for t in np.unique(probability)
    ])
    allowed = table[table["erfüllt P_min"]]
    return allowed.loc[allowed["F2"].idxmax(), "threshold"], table


candidates = {
    "logreg A": (scaler, logreg, FEATURES_A),
    "logreg B": (scaler_b, logreg_b, FEATURES_B),
    "logreg A balanced": (scaler_balanced, logreg_balanced, FEATURES_A),
    "logreg B balanced": (scaler_balanced_b, logreg_balanced_b, FEATURES_B),
}

THRESHOLDS = {}         # Modellname -> gewählte Schwelle, in Schritt 8 wieder gebraucht
threshold_tables = {}   # Modellname -> Scores für jede Schwelle, für den Plot
for name, (model_scaler, model, features) in candidates.items():
    probability = predict_probability(model_scaler, model, df_train_clean, features)
    THRESHOLDS[name], threshold_tables[name] = best_threshold(y_train, probability)
    train_results[f"{name} + Schwelle {THRESHOLDS[name]:.2f}"] = classification_scores(
        y_train, (probability >= THRESHOLDS[name]).astype(int))

pd.Series(THRESHOLDS, name="gewählte Schwelle").round(3)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 4.5), sharey=True)
for ax, name in zip(axes, ["logreg A", "logreg A balanced"]):
    table = threshold_tables[name]
    for score in ["F2", "precision", "recall"]:
        ax.plot(table["threshold"], table[score], label=score)
    ax.axhline(P_MIN, color="grey", linestyle="--", label=f"P_min = {P_MIN}")
    ax.axvline(0.5, color="lightgrey", linestyle="-", label="Standardschwelle 0,5")
    ax.axvline(THRESHOLDS[name], color="black", linestyle=":", label="gewählte Schwelle")
    ax.set_xlabel("Entscheidungsschwelle auf der vorhergesagten Wahrscheinlichkeit")
    ax.set_title(f"{name}: Scores für jede Schwelle (Training), gewählte Schwelle = {THRESHOLDS[name]:.2f}")
axes[0].set_ylabel("Score auf dem Trainings-Set")
fig.legend(*axes[0].get_legend_handles_labels(), loc="lower center", ncol=6)
plt.tight_layout(rect=(0, 0.08, 1, 1))
plt.show()
```

**Was wir im Plot sehen:** Beide Felder zeigen denselben Zielkonflikt wie beim Glucose-Cutoff: Eine niedrigere Schwelle markiert mehr Patientinnen, also steigt der Recall und die Precision sinkt. **Die Kurven des balanced-Modells sind nach rechts verschoben:** Seine Wahrscheinlichkeiten liegen insgesamt höher, es braucht also eine höhere Schwelle (0,27 statt 0,19) für dieselben Entscheidungen. Das ist der Punkt des Coaches: Nach dem Ändern der Gewichte passt die alte Schwelle nicht mehr.

### 7c. Alle Modelle auf dem Trainings-Set

```python
train_table = pd.DataFrame.from_dict(train_results, orient="index")
train_table.sort_values("F2", ascending=False).round(3)
```

**Was wir sehen:**

- **Die Schwelle ist der entscheidende Hebel.** Alle vier Modelle mit eigener Schwelle erreichen **F2 ≈ 0,80** und schlagen damit die Glucose-Regel (0,73) deutlich, bei Precision ≥ 0,5. Sie finden 91 bis 94 % der späteren Fälle. Die Modelle mit der Standardschwelle 0,5 bleiben unter der Regel.
- **Brauchen wir die Klassengewichte noch?** Kaum. Mit eigener Schwelle erreicht das Modell ohne Gewichte F2 0,795 (A) und 0,801 (B), mit Gewichten 0,799 und 0,803. Beide Hebel machen im Grunde dasselbe: Sie verschieben die Grenze zwischen „Diabetes“ und „kein Diabetes“. Sobald wir die Schwelle für unsere Metrik wählen, tragen die Gewichte wenig bei. Sie sorgen vor allem dafür, dass die Schwelle an einer „natürlicheren“ Stelle landet (0,27 statt 0,19).
- **A oder B?** Variante B (ohne Insulin und SkinThickness) ist mindestens so gut wie A. Das bestätigt die Koeffizienten aus 6.1: Das Modell braucht die beiden Spalten mit den meisten aufgefüllten Werten nicht.
- **Mit Vorsicht lesen:**
  - Die Unterschiede zwischen den vier Schwellen-Modellen liegen unter 0,01. Bei 614 Patientinnen sind das ein oder zwei Fälle, also **Rauschen, keine Rangfolge**.
  - Alle gewählten Schwellen liegen **genau an der Precision-Untergrenze** (Precision 0,50 bis 0,53). Bei neuen Patientinnen kann die Precision unter 0,5 fallen. Ob sie hält, zeigt das Test-Set in Schritt 8.
  - Alle Scores sind Trainingswerte, und die Schwellen wurden auf denselben Daten gewählt, sie sind also zu optimistisch.
- **Unser Kandidat für Schritt 8:** **logreg B balanced + Schwelle**, mit dem höchsten Trainings-F2 (0,80), den wenigsten Merkmalen und den Klassengewichten, die die Aufgabe verlangt. In Schritt 8 berichten wir trotzdem **alle** Modelle auf dem Test-Set und wechseln wegen der Testwerte nicht zu einem anderen Modell.

## Schritt 8: Test-Set vorbereiten und alle Modelle bewerten

**Jetzt, und erst jetzt, verwenden wir `df_test`.** Es steht für 154 neue Patientinnen, die kein Modell gesehen hat. Alles, was wir darauf anwenden, stammt aus den Trainingsdaten:

1. **Bereinigung** mit `apply_cleaning(df_test, TRAIN_MEDIANS)`: Der Schritt 0 → NaN lief schon vor dem Split, wir füllen die NaN also mit den **Trainings**-Medianen auf.
2. **Skalierung** mit den auf den Trainingsdaten gefitteten Scalern: nur `transform`, nie `fit`.
3. **Schwellen** aus `THRESHOLDS`, in Schritt 7 auf den Trainingsdaten gewählt.

**Keine Wahl mehr:** Unseren Kandidaten haben wir in Schritt 7 auf den Trainingsdaten gewählt (**logreg B balanced + Schwelle**). Wir berichten hier alle Modelle, wechseln aber wegen der Testwerte nicht zu einem anderen. Sonst hätte das Test-Set an der Entscheidung mitgewirkt und wäre kein unabhängiger Prüfstein mehr.

```python
df_test_clean = apply_cleaning(df_test, TRAIN_MEDIANS)

assert df_test_clean.isna().sum().sum() == 0, "im Test-Set gibt es noch NaN"
assert list(df_test_clean.columns) == list(df_train_clean.columns)

X_test = df_test_clean[FEATURES]
y_test = df_test_clean[TARGET]
print("X_test:", X_test.shape, "| Diabetes-Fälle im Test-Set:", y_test.sum())
print("Mit Trainings-Medianen aufgefüllte Werte je Spalte:")
df_test[ZERO_MEANS_MISSING].isna().sum()
```

**Eine Vorhersagefunktion pro Modell:** Jedes Modell muss seine Vorhersagen auf neuen Daten genau so treffen wie im Training: mit denselben Merkmalen, demselben Scaler und derselben Schwelle. Ein Dictionary hält die Namen identisch zur Trainingstabelle, und das `assert` prüft, dass kein Modell fehlt.

```python
predictors = {
    "Zufalls-Baseline": lambda data: dummy.predict(data[FEATURES]),
    "Referenz: alle haben Diabetes": lambda data: np.ones(len(data), dtype=int),
    f"Regel: Glucose >= {GLUCOSE_CUTOFF:.0f}": lambda data: rule_predict(data, GLUCOSE_CUTOFF),
    "logreg A (alle 8 Merkmale)": lambda data: logreg.predict(scaler.transform(data[FEATURES_A])),
    "logreg B (ohne Insulin, SkinThickness)": lambda data: logreg_b.predict(scaler_b.transform(data[FEATURES_B])),
    "logreg A balanced": lambda data: logreg_balanced.predict(scaler_balanced.transform(data[FEATURES_A])),
    "logreg B balanced": lambda data: logreg_balanced_b.predict(scaler_balanced_b.transform(data[FEATURES_B])),
}
for name, (model_scaler, model, features) in candidates.items():
    predictors[f"{name} + Schwelle {THRESHOLDS[name]:.2f}"] = (
        lambda data, s=model_scaler, m=model, f=features, t=THRESHOLDS[name]:
            (predict_probability(s, m, data, f) >= t).astype(int)
    )

assert predictors.keys() == train_results.keys(), "die Testmodelle müssen den Trainingsmodellen entsprechen"
```

```python
CHOSEN_MODEL = f"logreg B balanced + Schwelle {THRESHOLDS['logreg B balanced']:.2f}"

rows = {}
for name, predict in predictors.items():
    test_scores = classification_scores(y_test, predict(df_test_clean))
    rows[name] = {
        "F2 Training": train_results[name]["F2"], "F2 Test": test_scores["F2"],
        "precision Training": train_results[name]["precision"], "precision Test": test_scores["precision"],
        "recall Training": train_results[name]["recall"], "recall Test": test_scores["recall"],
        "accuracy Test": test_scores["accuracy"],
        "erfüllt P_min (Test)": test_scores["erfüllt P_min"],
    }

results = pd.DataFrame.from_dict(rows, orient="index")
results["F2 Training − Test"] = results["F2 Training"] - results["F2 Test"]
results.index = [f"► {name}" if name == CHOSEN_MODEL else name for name in results.index]
results.sort_values("F2 Training", ascending=False).round(3)
```

**Was wir sehen:**

- **Das gewählte Modell (►) hält:** F2 = 0,81 auf dem Test-Set, wie auf dem Trainings-Set (0,80). Die Precision bleibt über der Untergrenze (0,53 ≥ 0,5), der Recall liegt bei 0,93. Training und Test stimmen überein, es gibt also keinen Hinweis auf Overfitting. Die Schwelle, gewählt auf den Trainingsdaten, hat sich auf neue Patientinnen übertragen.
- **Es schlägt beide Baselines:** die Zufalls-Baseline (F2 0,38) weit, die Glucose-Regel (F2 0,76) knapper. Auf dem Test-Set schneidet die Regel besser ab als auf dem Trainings-Set, der Abstand schrumpft von 0,07 auf etwa 0,04.
- **Die Standardschwelle bricht ein:** Die Modelle ohne Klassengewichte mit Schwelle 0,5 fallen von F2 0,62 im Training auf etwa 0,50 im Test. Mit Klassengewichten oder eigener Schwelle sind die Modelle deutlich stabiler.
- **A und B sind auch auf dem Test-Set gleich:** Mit Klassengewichten treffen beide genau dieselben Entscheidungen. Insulin und SkinThickness ändern keine einzige Vorhersage.
- **Kleines Test-Set:** 54 Diabetes-Fälle. Ein Fall mehr oder weniger gefunden verschiebt den Recall um 0,019, F2 um etwa ebenso viel. Die vier Schwellen-Modelle (0,78 bis 0,81) sind also praktisch gleich. Der Abstand zur Glucose-Regel entspricht etwa vier zusätzlich gefundenen Fällen, ein echter, aber kleiner Vorteil.
- **Accuracy als Warnlampe:** Die nach F2 besten Modelle haben unter den logistischen Regressionen die *niedrigste* Accuracy (0,69 gegen 0,73). Wer nach Accuracy sortierte, würde für unseren Zweck das falsche Modell wählen.

### 8.1 Wo macht das beste Modell noch Fehler? Die Confusion Matrix

Zeilen = Wahrheit, Spalten = Vorhersage, 0 vor 1. `.ravel()` liefert TN, FP, FN, TP in dieser Reihenfolge. Zum Vergleich daneben die Glucose-Regel.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
for ax, name in zip(axes, [CHOSEN_MODEL, f"Regel: Glucose >= {GLUCOSE_CUTOFF:.0f}"]):
    matrix = confusion_matrix(y_test, predictors[name](df_test_clean))
    sns.heatmap(matrix, annot=True, fmt="d", cmap="Blues", cbar=False, ax=ax,
                xticklabels=["kein Diabetes (0)", "Diabetes (1)"], yticklabels=["kein Diabetes (0)", "Diabetes (1)"])
    ax.set_xlabel("Vorhersage")
    ax.set_ylabel("Wahrheit")
    ax.set_title(name)
plt.suptitle("Confusion Matrices auf dem Test-Set (154 Patientinnen)")
plt.tight_layout()
plt.show()

tn, fp, fn, tp = confusion_matrix(y_test, predictors[CHOSEN_MODEL](df_test_clean)).ravel()
print(f"{CHOSEN_MODEL}: TN = {tn}, FP = {fp}, FN = {fn}, TP = {tp}")
print(f"{tp} von {tp + fn} späteren Diabetes-Fällen gefunden, {fn} übersehen.")
print(f"{fp} Fehlalarme: {fp} von {tn + fp} Frauen ohne Diabetes bekämen eine unnötige Kontrolle.")
```

**Was wir sehen:**

- **Was das Modell richtig macht:** Es findet **50 der 54** Frauen, die innerhalb von fünf Jahren Diabetes entwickeln (TP). Nur **4 werden übersehen** (FN): Das ist der Fehler, den wir zu Beginn als den teuren festgelegt haben, und das Modell hält ihn klein. Die Glucose-Regel übersieht 8.
- **Wo es noch Fehler macht:** **44 Fehlalarme** (FP). 44 der 100 Frauen, die gesund bleiben, würden zu einer unnötigen Kontrolle oder Beratung eingeladen. Von allen 94 Alarmen sind 50 echte Fälle: Precision 0,53, also ist etwa jeder zweite Alarm richtig, wie es unser `P_MIN` verlangt hat.
- **Ist das ein guter Tausch?** Nach unserer eigenen Annahme (ein übersehener Fall wiegt etwa so viel wie vier Fehlalarme) ja: Das Modell tauscht 4 übersehene Fälle weniger als die Regel (4 statt 8) gegen 5 Fehlalarme mehr (44 statt 39). Bei einem Screening zur Vorbeugung, wo ein Fehlalarm „nur“ eine Kontrolle kostet, ist das vertretbar. Bedeutete ein Fehlalarm eine invasive Untersuchung, müssten wir β und `P_MIN` anders wählen.
- **Vergleich mit dem Paper:** Das ADAP-Netz im Paper erreichte Sensitivität = Spezifität = 0,76. Unser Modell arbeitet an einer anderen Stelle des Zielkonflikts: Sensitivität (Recall) 0,93, Spezifität 56 / 100 = 0,56. Die Zahlen sind nicht direkt vergleichbar (anderer Split, anderes Modell, anderer Arbeitspunkt), aber sie zeigen dasselbe Bild: Die Daten tragen ein deutliches, aber begrenztes Signal.

**Forschungsfrage:** *Welche bei der Untersuchung gemessenen Merkmale zeigen an, dass eine Pima-Frau in den nächsten fünf Jahren Diabetes entwickeln wird?*

- **Das Modell:** Eine logistische Regression mit Klassengewichten und einer für F2 gewählten Schwelle findet bei neuen Patientinnen 93 % der späteren Fälle, bei einer Precision von 0,53. Sie schlägt beide Baselines und hält ihre Trainingswerte auf dem Test-Set.

- β = 2 und `P_MIN = 0.5` sind unsere Annahmen über die Kosten. Bei anderen Kosten wäre ein anderes Modell mit einer anderen Schwelle „das beste“.
- 768 Frauen aus einer Bevölkerungsgruppe (Pima, mindestens 21 Jahre alt): Die Ergebnisse lassen sich nicht ohne Weiteres auf andere Gruppen übertragen.
